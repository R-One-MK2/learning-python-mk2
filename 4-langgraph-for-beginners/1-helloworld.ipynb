```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    message: str


def greeting_node(state: AgentState) -> AgentState:
    state["message"] = "Hey " + state["message"] + ", Hows your day going?"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeter", greeting_node)
graph.set_entry_point("greeter")
graph.set_finish_point("greeter")

app = graph.compile()
```

```python
from IPython.display import Image, display

display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"message": "Bob"})
```

```python
class ComplimentState(TypedDict):
    name: str


def give_compliments(state: ComplimentState) -> ComplimentState:
    state["name"] = (
        state["name"] + ", you are doing an amazhing job learning Langgraph!"
    )
    return state
```

```python
compliment_graph = StateGraph(ComplimentState)

compliment_graph.add_node("complimenter", give_compliments)
compliment_graph.set_entry_point("complimenter")  # start here
compliment_graph.set_finish_point("complimenter")  # start here

compliment_app = compliment_graph.compile()
```

```python
from IPython.display import Image, display

display(Image(compliment_app.get_graph().draw_mermaid_png()))
```

```python
compliment = compliment_app.invoke({"name": "Bob"})
```
